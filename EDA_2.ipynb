```python
#14
np.random.seed(1)
b = np.random.randint(0,10,size=(2,3))
print(b)
a= np.random.randint(0,10,size=(2,3))
print(a)
v_stack = np.vstack((b,a))
print(v_stack)
v_concat=np.concatenate((b,a),axis=0)
print(v_concat)
h_stack = np.hstack((b,a))
print(h_stack)
h_concat=np.concatenate((b,a),axis=1)
print(h_concat)
```

```python
#16
np.random.seed(1)
x = np.random.randint(0,100,size=(6,4))
print("Original Array\n",x)
h1,h2=np.hsplit(x,2)
print("HSplit1\n",h1)
print("HSplit2\n",h2)
v1,v2 = np.vsplit(x,2)
print("VSplit1\n",v1)
print("VSplit2\n",v2)
```
