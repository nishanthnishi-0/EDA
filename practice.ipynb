```python
import numpy as np
import pandas as pd
```

```python
index_tuples = [
    ("Karnataka", 2010),
    ("Karnataka", 2020),
    ("Maharashtra", 2010),
    ("Maharashtra", 2020),
    ("Tamil Nadu", 2010),
    ("Tamil Nadu", 2020),
]
multi_index = pd.MultiIndex.from_tuples(index_tuples, names=["State", "Year"])
populations = [100, 200, 300, 400, 500, 600]
population_series = pd.Series(populations, index=multi_index, name="Population (M)")
print(population_series)
```

```python
s = pd.Series([1, np.nan, 2, None, 3, np.nan])
zero = s.fillna(0)
ffill = s.ffill()
filled_bfill = s.bfill()
print(zero)
print(ffill)
print(filled_bfill)
```

```python
departments = ["CSE", "AIML"]
years = [2023, 2024]
np.random.seed(1)
multi_index = pd.MultiIndex.from_product(
    [departments, years], names=["Department", "Year"]
)
data = np.random.randint(50, 100, size=(4, 2))
df = pd.DataFrame(data, index=multi_index, columns=["Mathematics", "Science"])
print(df)
```
